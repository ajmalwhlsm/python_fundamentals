```python
def check_season(month):
  month=month.lower().strip()
  if month in ['december','januray','february']:
    print('winter')
  elif month in ['march','april','may']:
    print('spring')
  elif month in ['june','july','august']:
    print('summer')
  elif month in ['september','october','november']:
    print('autumn')
  else:
    print("invalid month")
check_season('april')
```

```python
import random
import string
def list_of_hexa_colors(tim):

  letters=string.ascii_letters[0:6]
  digits=string.digits
  chr=letters+digits
  loh=[]
  for iteration in range(tim):
    loh.append('#'+''.join(random.choice(chr) for l in range(6)))
  return loh
list_of_hexa_colors(5)
```

```python
import random
def list_of_rgb_colors(tim):
  log=[]
  for times in range(tim):

   r=random.randint(0,255)
   g=random.randint(0,255)
   b=random.randint(0,255)

   log.append(f"rgb({r},{g},{b})")
  return log
list_of_rgb_colors(6)
```

```python
def generate_colors(type,num):
  if type=='hexa':
    return(list_of_hexa_colors(num))
  elif type=='rgb':
    return(list_of_rgb_colors(num))
```

```python
def start_letter(contri):
  contri.sort(reverse=False)
  result={}
  for country in contri:
    first_letter=country[0]
    if first_letter not in result:
      result[first_letter]=1
    else:
      result[first_letter]+=1
  return result

```
